# 🎯 Modul 02 Hands-on: Classical ML Algorithms Deep Dive

Pada hands-on ini, Anda akan membandingkan 4 algoritma klasik:
1. **Logistic Regression**
2. **K-Nearest Neighbors (KNN)**
3. **Gaussian Naive Bayes**
4. **The Sabotage Lab: The Curse of k=1 in KNN**

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

X, y = make_moons(n_samples=300, noise=0.25, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print("Dataset non-linear siap!")
```

## 1. The Sabotage Lab: KNN k=1 (Extreme Overfitting) vs k=15
Dengan k=1, model menghafal setiap titik noise training data, menghasilkan boundary pulau-pulau kecil.

```python
knn_1 = KNeighborsClassifier(n_neighbors=1).fit(X_train, y_train)
knn_15 = KNeighborsClassifier(n_neighbors=15).fit(X_train, y_train)

print(f"KNN (k=1)  -> Train Acc: {knn_1.score(X_train, y_train)*100:.1f}% | Test Acc: {knn_1.score(X_test, y_test)*100:.1f}%")
print(f"KNN (k=15) -> Train Acc: {knn_15.score(X_train, y_train)*100:.1f}% | Test Acc: {knn_15.score(X_test, y_test)*100:.1f}%")

# Plot Decision Boundary
h = 0.05
x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))

fig, axes = plt.subplots(1, 2, figsize=(11, 5))
for ax, model, title in zip(axes, [knn_1, knn_15], ["SABOTASE: KNN k=1 (Overfit)", "BENAR: KNN k=15 (Smooth)"]):
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.Spectral)
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.Spectral, edgecolors="k", s=30)
    ax.set_title(title, fontweight="bold")
plt.tight_layout(); plt.show()
```
