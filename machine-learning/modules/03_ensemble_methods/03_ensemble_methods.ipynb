# 🌳 Modul 03 Hands-on: Tree-Based Models & Ensemble Methods

Pada hands-on ini, Anda akan membuktikan kekuatan model berbasis pohon:
1. **Single Decision Tree vs Random Forest vs Gradient Boosting**.
2. **The Sabotage Lab**: Membiarkan pohon tumbuh tanpa batas (`max_depth=None`).
3. **Visualisasi Feature Importance**.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score

X, y = make_classification(n_samples=800, n_features=12, n_informative=8, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
print("Dataset tabular klasifikasi siap!")
```

## 1. The Sabotage Lab: Decision Tree Tanpa Batas vs Ensembles

```python
# Sabotase: Single tree tanpa batas kedalaman
tree_unlimited = DecisionTreeClassifier(max_depth=None, random_state=42).fit(X_train, y_train)
rf = RandomForestClassifier(n_estimators=100, max_depth=6, random_state=42).fit(X_train, y_train)
gb = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=42).fit(X_train, y_train)

print(f"Tree Unlimited -> Train: {tree_unlimited.score(X_train, y_train)*100:.1f}% | Test: {tree_unlimited.score(X_test, y_test)*100:.1f}% (Overfit Gap!)")
print(f"Random Forest  -> Train: {rf.score(X_train, y_train)*100:.1f}% | Test: {rf.score(X_test, y_test)*100:.1f}%")
print(f"Grad Boosting  -> Train: {gb.score(X_train, y_train)*100:.1f}% | Test: {gb.score(X_test, y_test)*100:.1f}%")
```

## 2. Feature Importance Analysis

```python
importances = rf.feature_importances_
indices = np.argsort(importances)[::-1]

plt.figure(figsize=(9, 4))
plt.bar(range(X.shape[1]), importances[indices], color="forestgreen")
plt.title("Random Forest Feature Importance Ranking", fontsize=12, fontweight="bold")
plt.xlabel("Index Fitur"); plt.ylabel("Gini Importance")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
